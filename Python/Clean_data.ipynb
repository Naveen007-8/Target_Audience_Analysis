```python
tran = tran.merge(credit[['Cust_Id','Credit_Limit']], on='Cust_Id', how='left')
```

```python
Q1 = tran['Tran_Amount'].quantile(0.25)
Q2 = tran['Tran_Amount'].quantile(0.75)

IQR = Q2 - Q1

lower = Q1 - 1.5 * IQR
upper = Q2 + 1.5 * IQR
```

```python
tran['Tran_Amount'] = tran['Tran_Amount'].astype('float64')
```

```python
condition = (tran['Tran_Amount'] < lower) | (tran['Tran_Amount'] > upper)

tran.loc[condition,'Tran_Amount'] = tran.groupby(['Platform','Credit_Limit'])['Tran_Amount'].transform('median')
```

```python
tran.describe()
```

```python
(tran['Tran_Amount'] == 0).sum()
```

```python
tran.loc[tran['Tran_Amount'] == 0, 'Tran_Amount'] = np.nan
```

```python
tran['Tran_Amount'] = tran['Tran_Amount'].fillna(
    tran.groupby(['Platform','Credit_Limit'])['Tran_Amount'].transform('median')
)
```
